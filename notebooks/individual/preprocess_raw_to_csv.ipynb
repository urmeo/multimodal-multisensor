# Convert individual source exports

Convert the individual raw exports to CSV while keeping source values and row order. These files differ from the case-study recordings, but their participant mapping is undocumented. Write derived eye samples and one row per low-movement event to a separate output directory.

Read from `MMS_DATA_ROOT` or the checkout's `data/` directory. Optional exports require `MMS_OUTPUT_ROOT` outside the source data.


```python
from mms import exploration as ex, io

SOURCE = "individual"
```

```python
for sensor in ("hr", "ibi", "sed"):
    for session in ex.SESSIONS:
        ex.convert_raw(sensor, session, source=SOURCE)
```

```python
for session in (1, 2, 3):
    frame = io.load_raw("sed", session, source=SOURCE, data_root=ex.data_root())
    samples, events = ex.derive_eye(frame)
    ex.export_frame(samples, f"individual/derived/eye_samples_{session:02d}.csv")
    ex.export_frame(events, f"individual/derived/low_movement_events_{session:02d}.csv")
```
