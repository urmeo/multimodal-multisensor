# HADS item intervals and interval samples

Summarize interval samples during HADS item intervals when acquisition timezone metadata is available. Keep channels separate and break adjacent differences at invalid samples. Raw answer codes lack verified scoring keys, so these summaries do not classify anxiety or apply decrease thresholds.

Read from `MMS_DATA_ROOT` or the checkout's `data/` directory. Optional exports require `MMS_OUTPUT_ROOT` outside the source data.


```python
from mms import exploration as ex

SOURCE = "individual"

intervals = ex.recordings("ibi", source=SOURCE)
interval_table = ex.interval_summary(intervals)
print(interval_table)
features = ex.aligned_questions(source=SOURCE)
if not features.empty:
    metrics = ["mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]
    print(features.groupby(["Session", "Type", "Channel"])[metrics].mean())
    hads = features.loc[features.Type.eq("HADS")]
    print(hads.groupby(["Session", "Channel"])[["interval_sd_ms", "interval_rmsdiff_ms"]].mean())
    ex.export_frame(hads, "individual/derived/hads_interval_item_features.csv")
```
