# Exploratory item-level associations

Describe item-level associations within each instrument and channel. Question containment requires acquisition timezone metadata; separate recordings cannot be aligned by row index. Report Spearman coefficients and finite pair counts without inferential p values for repeated within-person observations.

Raw answer codes lack verified scoring keys. Least-squares fits describe the same data used to fit them and do not establish prediction or diagnosis.

Use `MMS_DATA_ROOT` for another dataset and `MMS_OUTPUT_ROOT` for exports outside source data.


```python
import matplotlib.pyplot as plt
from mms import exploration as ex

SOURCE = "individual"

heart = ex.recordings("hr", source=SOURCE)
hr_table = ex.hr_summary(heart)
print(hr_table)
ex.plot_traces(heart, "hr")
```

```python
features = ex.aligned_questions(source=SOURCE)
if not features.empty:
    metrics = ["mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]
    print(features.groupby(["Session", "Type", "Channel"])[metrics].mean())
    for (kind, channel), part in features.groupby(["Type", "Channel"]):
        plt.scatter(part["response_s"], part["mean_hr_bpm"], s=12)
        plt.title(f"{kind}, channel {channel}: within-item descriptive association")
        plt.xlabel("Response duration (s)")
        plt.ylabel("Mean HR (bpm)")
        plt.show()
        plt.close()
```

```python
for (kind, channel), result in ex.question_correlations(features).items():
    print(f"{kind}, channel {channel}: repeated items; exploratory association only.")
    ex.plot_correlations(result)
```

```python
if not features.empty:
    metrics = [
        "response_s", "mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"
    ]
    print(features.groupby(["Type", "Channel"])[metrics].describe())
```
