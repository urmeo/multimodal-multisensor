# HR and interval recordings

Compare heart-rate and interval samples by channel. Reuse the one recorded baseline for descriptive differences; separate baselines for each visit are unavailable. Question alignment requires a documented acquisition timezone and clock synchronization confirmed by acquisition records. Neither is available here, so question-aligned summaries are skipped. The analysis does not apply clinical heart-rate or interval cutoffs.

Use `MMS_DATA_ROOT` for another dataset and `MMS_OUTPUT_ROOT` for exports outside source data.


```python
from mms import exploration as ex

SOURCE = "individual"

heart = ex.recordings("hr", source=SOURCE)
hr_table = ex.hr_summary(heart)
print(hr_table)
ex.plot_summary(hr_table, "Mean HR (bpm)")
print(ex.baseline_changes(hr_table, ["Mean HR (bpm)"]))
```

```python
intervals = ex.recordings("ibi", source=SOURCE)
interval_table = ex.interval_summary(intervals)
print(interval_table)
ex.plot_summary(interval_table, "Adjacent-sample RMS difference (ms)")
metrics = ["Sample SD (ms)", "Adjacent-sample RMS difference (ms)"]
print(ex.baseline_changes(interval_table, metrics))
ex.plot_traces(intervals, "ibi")
```

```python
features = ex.aligned_questions(source=SOURCE)
if not features.empty:
    metrics = ["mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]
    print(features.groupby(["Session", "Type", "Channel"])[metrics].mean())
```
