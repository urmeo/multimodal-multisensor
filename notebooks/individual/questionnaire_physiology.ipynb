# Raw responses and descriptive physiology

Summarize all 88 questionnaire items, requiring each item identity exactly once. Stored answer codes lack verified option and reversal keys. HADS subscale totals and universal STAI/FQ categories are therefore unavailable. Keep raw codes separate by instrument, without radar normalization or severity labels. Physiology joins require a documented acquisition timezone and clock synchronization confirmed by acquisition records. Neither is available here, so these joins are skipped.

Use `MMS_DATA_ROOT` for another dataset and `MMS_OUTPUT_ROOT` for exports outside source data.


```python
import matplotlib.pyplot as plt
from mms import exploration as ex

SOURCE = "individual"

heart = ex.recordings("hr", source=SOURCE)
hr_table = ex.hr_summary(heart)
print(hr_table)
intervals = ex.recordings("ibi", source=SOURCE)
interval_table = ex.interval_summary(intervals)
print(interval_table)
eyes = ex.recordings("eye", source=SOURCE)
eye_table = ex.eye_summary(eyes)
print(eye_table)
features = ex.aligned_questions(source=SOURCE)
if not features.empty:
    metrics = ["mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]
    print(features.groupby(["Session", "Type", "Channel"])[metrics].mean())
```

```python
durations, raw_codes = ex.response_tables(source=SOURCE)
print(durations)
print(raw_codes)
```

```python
for kind, part in raw_codes.groupby("Type"):
    plt.plot(part.Session, part.mean_raw_code, "o-", label=kind)
plt.title("Mean raw codes, separate instruments; not keyed scale scores")
plt.ylabel("Mean stored answer code")
plt.legend()
plt.show()
plt.close()
```
