# Within-question response durations

Summarize validated question start-to-answer durations and compare them with `Time(s)`. Means and standard deviations are in seconds. Repeated items are observations within a person, not independent participants.

Use `MMS_DATA_ROOT` for another dataset and `MMS_OUTPUT_ROOT` for exports outside source data.


```python
import matplotlib.pyplot as plt
from mms import exploration as ex

SOURCE = "individual"

durations, raw_codes = ex.response_tables(source=SOURCE)
print(durations)
print(raw_codes)
```

```python
for kind, part in durations.groupby("Type"):
    plt.plot(part.Session, part.mean_response_s, "o-", label=kind)
plt.ylabel("Mean response duration (s)")
plt.legend()
plt.show()
plt.close()
```
