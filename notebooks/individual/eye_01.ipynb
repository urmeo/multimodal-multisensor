# Eye recording summaries

Summarize pupil readings and closure candidates by session for the separate individual recording set.

The individual participant mapping is undocumented. Pupil quality must be at least 0.5. Low-movement runs use source-vector steps below 0.01 and gaps at most 0.1 seconds. Closure candidates require both openness values at most 1, quality at least 0.5, observed opening boundaries, durations of 0.05–0.5 seconds, and gaps at most 0.1 seconds. These are exploratory rules, without validated blink or anxiety interpretation.

Use `MMS_DATA_ROOT` for another dataset and `MMS_OUTPUT_ROOT` for exports outside source data.


```python
from mms import exploration as ex

SOURCE = "individual"

eyes = ex.recordings("eye", source=SOURCE)
eye_table = ex.eye_summary(eyes)
print(eye_table)
print(ex.baseline_changes(eye_table, ["Mean pupil (source units)", "Closure candidates/min"]))
print(ex.gaze_rate_summary({"Baseline": eyes["Baseline"]}))
ex.plot_summary(eye_table, "Mean pupil (source units)")
ex.plot_summary(eye_table, "Closure candidates/min")
ex.plot_traces(eyes, "eye")
```

```python
features = ex.aligned_questions(source=SOURCE)
if not features.empty:
    metrics = ["mean_hr_bpm", "interval_rmsdiff_ms", "mean_pupil_source_units"]
    print(features.groupby(["Session", "Type", "Channel"])[metrics].mean())
```
