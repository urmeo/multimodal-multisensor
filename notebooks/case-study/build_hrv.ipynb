# Build scratch interval variability

Calculate trailing variability from 30 interval samples per channel, with at most 29 internal differences. Keep source rows and invalid positions. Write derived files to a separate output directory; historical HRV files remain unchanged.

Read from `MMS_DATA_ROOT` or the checkout's `data/` directory. Optional exports require `MMS_OUTPUT_ROOT` outside the source data.


```python
from mms import exploration as ex, hrv, io

SOURCE = "case-study"

for session in ex.SESSIONS:
    frame = io.load_ibi(session, source=SOURCE, data_root=ex.data_root())
    derived = hrv.hrv_rolling(frame, window_beats=30)
    suffix = "" if session is None else f"_{session:02d}"
    ex.export_frame(derived, f"{SOURCE}/derived/interval_variability{suffix}.csv")
```
