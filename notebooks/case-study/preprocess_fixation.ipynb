# Build scratch low-movement events

Keep all eye samples and quality gaps, then write one row per low-movement event.

Read from `MMS_DATA_ROOT` or the checkout's `data/` directory. Optional exports require `MMS_OUTPUT_ROOT` outside the source data.


```python
from mms import exploration as ex, io

SOURCE = "case-study"

for session in ex.SESSIONS:
    frame = io.load_raw("sed", session, source=SOURCE, data_root=ex.data_root())
    samples, events = ex.derive_eye(frame)
    suffix = "" if session is None else f"_{session:02d}"
    ex.export_frame(samples, f"{SOURCE}/derived/eye_samples{suffix}.csv")
    ex.export_frame(events, f"{SOURCE}/derived/low_movement_events{suffix}.csv")
```
