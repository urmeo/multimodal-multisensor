# Heart-rate sample variability

Compare heart-rate standard deviations by channel and session. Use finite positive readings with confidence equal to 1. The samples are autocorrelated, so independent-sample population tests would be inappropriate.

Read from `MMS_DATA_ROOT` or the checkout's `data/` directory. Optional exports require `MMS_OUTPUT_ROOT` outside the source data.


```python
from mms import exploration as ex

SOURCE = "case-study"

heart = ex.recordings("hr", source=SOURCE)
hr_table = ex.hr_summary(heart)
print(hr_table)
ex.plot_summary(hr_table, "HR SD (bpm)")
```

```python
print(ex.baseline_changes(hr_table, ["Mean HR (bpm)", "HR SD (bpm)"]))
```
