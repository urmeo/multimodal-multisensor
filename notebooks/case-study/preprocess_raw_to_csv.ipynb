# Convert raw exports to scratch CSVs

Convert each semicolon-delimited source export to CSV while preserving its values. Set `MMS_OUTPUT_ROOT` to a directory outside the source data to write copies. Without it, conversion skips the write.

Read from `MMS_DATA_ROOT` or the checkout's `data/` directory. Optional exports require `MMS_OUTPUT_ROOT` outside the source data.


```python
from mms import exploration as ex

SOURCE = "case-study"

for sensor in ("hr", "ibi", "sed"):
    for session in ex.SESSIONS:
        ex.convert_raw(sensor, session, source=SOURCE)
```
