```python
import pandas as pd

DATA = '../data/individual/processed'

# load baseline HR
baseline_hr = pd.read_csv(f'{DATA}/hr.csv')

# load HR data
hr_01 = pd.read_csv(f'{DATA}/hr_01.csv')
hr_02 = pd.read_csv(f'{DATA}/hr_02.csv')
hr_03 = pd.read_csv(f'{DATA}/hr_03.csv')

# clean HR data
def clean_hr_data(hr_data):
    return hr_data[hr_data['confidence'] == 1.0].copy()

baseline_hr_clean = clean_hr_data(baseline_hr)
hr_01_clean = clean_hr_data(hr_01)
hr_02_clean = clean_hr_data(hr_02)
hr_03_clean = clean_hr_data(hr_03)

# average per session
baseline_avg_hr = baseline_hr_clean['heart_rate'].mean()
avg_hr_01 = hr_01_clean['heart_rate'].mean()
avg_hr_02 = hr_02_clean['heart_rate'].mean()
avg_hr_03 = hr_03_clean['heart_rate'].mean()

# baseline difference
hr_difference_01 = avg_hr_01 - baseline_avg_hr
hr_difference_02 = avg_hr_02 - baseline_avg_hr
hr_difference_03 = avg_hr_03 - baseline_avg_hr

# anxiety threshold
hr_anxiety_threshold = 100

# check HR anxiety
anxiety_hr_baseline = baseline_avg_hr > hr_anxiety_threshold
anxiety_hr_01 = avg_hr_01 > hr_anxiety_threshold
anxiety_hr_02 = avg_hr_02 > hr_anxiety_threshold
anxiety_hr_03 = avg_hr_03 > hr_anxiety_threshold

# extract for correlation
baseline_hr_values = baseline_hr_clean['heart_rate']
hr_01_values = hr_01_clean['heart_rate']
hr_02_values = hr_02_clean['heart_rate']
hr_03_values = hr_03_clean['heart_rate']

hr_data_for_corr = pd.DataFrame({
    'Baseline': baseline_hr_values.reset_index(drop=True),
    'Test_01': hr_01_values.reset_index(drop=True),
    'Test_02': hr_02_values.reset_index(drop=True),
    'Test_03': hr_03_values.reset_index(drop=True)
})

correlation_matrix = hr_data_for_corr.corr(method='spearman')

results = {
    "Baseline Average HR": baseline_avg_hr,
    "Baseline Anxiety": "Anxiety" if anxiety_hr_baseline else "Normal",
    "Session 1 Average HR": avg_hr_01,
    "Session 1 Anxiety": "Anxiety" if anxiety_hr_01 else "Normal",
    "Difference from Baseline in Session 1": hr_difference_01,
    "Session 2 Average HR": avg_hr_02,
    "Session 2 Anxiety": "Anxiety" if anxiety_hr_02 else "Normal",
    "Difference from Baseline in Session 2": hr_difference_02,
    "Session 3 Average HR": avg_hr_03,
    "Session 3 Anxiety": "Anxiety" if anxiety_hr_03 else "Normal",
    "Difference from Baseline in Session 3": hr_difference_03
}

results_df = pd.DataFrame([results])
print(results_df)

print("Spearman's Correlation Coefficient Matrix:")
print(correlation_matrix)
```

```python
import pandas as pd
from scipy.stats import pearsonr
import matplotlib.pyplot as plt
import seaborn as sns

DATA = '../data/individual/processed'
PSY = '../data/individual/psychometric'

# load data
hr_01 = pd.read_csv(f'{DATA}/hr_01.csv')
hr_02 = pd.read_csv(f'{DATA}/hr_02.csv')
hr_03 = pd.read_csv(f'{DATA}/hr_03.csv')

psychometric_01 = pd.read_csv(f'{PSY}/Psychometric_Test_Results_01.csv')
psychometric_02 = pd.read_csv(f'{PSY}/Psychometric_Test_Results_02.csv')
psychometric_03 = pd.read_csv(f'{PSY}/Psychometric_Test_Results_03.csv')

# high confidence only
hr_01 = hr_01[hr_01['confidence'] == 1.0]
hr_02 = hr_02[hr_02['confidence'] == 1.0]
hr_03 = hr_03[hr_03['confidence'] == 1.0]

# convert datetimes
for df in [hr_01, hr_02, hr_03]:
    df['datetime'] = pd.to_datetime(df['datetime'], utc=True).dt.tz_convert(None)

for df in [psychometric_01, psychometric_02, psychometric_03]:
    df['Question Start Time'] = pd.to_datetime(df['Question Start Time'], utc=True).dt.tz_convert(None)
    df['Question Answer Time'] = pd.to_datetime(df['Question Answer Time'], utc=True).dt.tz_convert(None)

# answer duration
psychometric_01['answer_duration'] = (psychometric_01['Question Answer Time'] - psychometric_01['Question Start Time']).dt.total_seconds()
psychometric_02['answer_duration'] = (psychometric_02['Question Answer Time'] - psychometric_02['Question Start Time']).dt.total_seconds()
psychometric_03['answer_duration'] = (psychometric_03['Question Answer Time'] - psychometric_03['Question Start Time']).dt.total_seconds()

# merge HR + psychometric
merged_01 = pd.merge_asof(hr_01.sort_values('datetime'), psychometric_01[['Question Start Time', 'answer_duration', 'Answer']].sort_values('Question Start Time'),
                          left_on='datetime', right_on='Question Start Time', direction='nearest', tolerance=pd.Timedelta('30s'))
merged_02 = pd.merge_asof(hr_02.sort_values('datetime'), psychometric_02[['Question Start Time', 'answer_duration', 'Answer']].sort_values('Question Start Time'),
                          left_on='datetime', right_on='Question Start Time', direction='nearest', tolerance=pd.Timedelta('30s'))
merged_03 = pd.merge_asof(hr_03.sort_values('datetime'), psychometric_03[['Question Start Time', 'answer_duration', 'Answer']].sort_values('Question Start Time'),
                          left_on='datetime', right_on='Question Start Time', direction='nearest', tolerance=pd.Timedelta('30s'))

# drop NaN rows
merged_01_clean = merged_01.dropna(subset=['heart_rate', 'answer_duration', 'Answer'])
merged_02_clean = merged_02.dropna(subset=['heart_rate', 'answer_duration', 'Answer'])
merged_03_clean = merged_03.dropna(subset=['heart_rate', 'answer_duration', 'Answer'])

# correlations per session
correlation_01_duration = pearsonr(merged_01_clean['heart_rate'], merged_01_clean['answer_duration'])[0]
correlation_01_score = pearsonr(merged_01_clean['heart_rate'], pd.to_numeric(merged_01_clean['Answer'], errors='coerce'))[0]

correlation_02_duration = pearsonr(merged_02_clean['heart_rate'], merged_02_clean['answer_duration'])[0]
correlation_02_score = pearsonr(merged_02_clean['heart_rate'], pd.to_numeric(merged_02_clean['Answer'], errors='coerce'))[0]

correlation_03_duration = pearsonr(merged_03_clean['heart_rate'], merged_03_clean['answer_duration'])[0]
correlation_03_score = pearsonr(merged_03_clean['heart_rate'], pd.to_numeric(merged_03_clean['Answer'], errors='coerce'))[0]

correlations_extended = pd.DataFrame({
    'Session': ['Session 01', 'Session 02', 'Session 03'],
    'Correlation Duration': [correlation_01_duration, correlation_02_duration, correlation_03_duration],
    'Correlation Score': [correlation_01_score, correlation_02_score, correlation_03_score]
})

print("Heart Rate vs. Answer Duration and Answer Score Correlations")
print(correlations_extended)

# compare across sessions
psychometric_01['Session'] = 'Session 01'
psychometric_02['Session'] = 'Session 02'
psychometric_03['Session'] = 'Session 03'

all_sessions = pd.concat([psychometric_01, psychometric_02, psychometric_03])

plt.figure(figsize=(12, 6))
sns.boxplot(data=all_sessions, x='Session', y='answer_duration')
plt.title('Comparison of Answer Duration Across Sessions')
plt.ylabel('Answer Duration (seconds)')
plt.xlabel('Session')
plt.show()
plt.close()

answer_duration_stats = all_sessions[['Session', 'answer_duration']].groupby('Session').describe()
print("Comparison of Answer Duration Across Sessions")
print(answer_duration_stats)
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

DATA = '../data/individual/processed'

# merge eye + HRV data
qq2 = pd.read_csv(f'{DATA}/QQ2.csv')
qqhrv = pd.read_csv(f'{DATA}/QQHRV.csv')

merged = qq2.merge(qqhrv, on=['Test', 'Type', 'Start Time', 'End Time'],
                   how='inner', suffixes=('', '_hrv'))

# biometric columns
bio_cols = ['Average Pupil Dilation', 'Average Left Blink Rate', 'Average Right Blink Rate', 'RMSSD', 'SDNN']
available = [c for c in bio_cols if c in merged.columns]
bio_data = merged[available].dropna()

n = len(available)
r_matrix = np.zeros((n, n))
p_matrix = np.zeros((n, n))

# Spearman correlations
for i in range(n):
    for j in range(n):
        r, p = stats.spearmanr(bio_data[available[i]], bio_data[available[j]])
        r_matrix[i, j] = r
        p_matrix[i, j] = p

r_df = pd.DataFrame(r_matrix, index=available, columns=available)
p_df = pd.DataFrame(p_matrix, index=available, columns=available)

# annotate with stars
annot = np.empty_like(r_matrix, dtype=object)
for i in range(n):
    for j in range(n):
        stars = "***" if p_df.iloc[i, j] < 0.001 else "**" if p_df.iloc[i, j] < 0.01 else "*" if p_df.iloc[i, j] < 0.05 else ""
        annot[i, j] = f"{r_df.iloc[i, j]:.2f}{stars}"

# heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(r_df, annot=annot, fmt='', cmap='coolwarm', vmin=-1, vmax=1, center=0)
plt.title('Cross-Modal Spearman Correlation\n(* p<.05, ** p<.01, *** p<.001)')
plt.tight_layout()
plt.show()
plt.close()

print(f"\nMerged {len(merged)} question-level observations across eye + HRV modalities")
print(f"Available biometric features: {available}")
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

DATA = '../data/individual/processed'

# reload merged data
qq2 = pd.read_csv(f'{DATA}/QQ2.csv')
qqhrv = pd.read_csv(f'{DATA}/QQHRV.csv')
merged = qq2.merge(qqhrv, on=['Test', 'Type', 'Start Time', 'End Time'],
                   how='inner', suffixes=('', '_hrv'))

bio_cols = ['Average Pupil Dilation', 'Average Left Blink Rate', 'Average Right Blink Rate', 'RMSSD', 'SDNN']
available = [c for c in bio_cols if c in merged.columns]

# per-session correlations
print("=== Per-Session Cross-Modal Correlations ===\n")
test_col = 'Test'
for session in sorted(merged[test_col].unique()):
    session_data = merged[merged[test_col] == session][available].dropna()
    if len(session_data) < 5:
        continue
    print(f"--- {session} (n={len(session_data)}) ---")
    for i in range(len(available)):
        for j in range(i+1, len(available)):
            r, p = stats.spearmanr(session_data[available[i]], session_data[available[j]])
            if p < 0.05:
                print(f"  {available[i]} vs {available[j]}: r={r:.3f}, p={p:.4f} *")
    print()

# scatter matrix
fig, axes = plt.subplots(len(available), len(available), figsize=(14, 14))
for i in range(len(available)):
    for j in range(len(available)):
        ax = axes[i, j]
        if i == j:
            ax.hist(merged[available[i]].dropna(), bins=20, alpha=0.7, color='steelblue')
        else:
            ax.scatter(merged[available[j]], merged[available[i]], alpha=0.3, s=10)
        if i == len(available) - 1:
            ax.set_xlabel(available[j], fontsize=7, rotation=45)
        if j == 0:
            ax.set_ylabel(available[i], fontsize=7)
        ax.tick_params(labelsize=6)

plt.suptitle('Cross-Modal Scatter Matrix', y=1.01)
plt.tight_layout()
plt.show()
plt.close()
```

```python
import pandas as pd
import numpy as np
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

DATA = '../data/individual/processed'

# load merged data
qq2 = pd.read_csv(f'{DATA}/QQ2.csv')
qqhrv = pd.read_csv(f'{DATA}/QQHRV.csv')
merged = qq2.merge(qqhrv, on=['Test', 'Type', 'Start Time', 'End Time'],
                   how='inner', suffixes=('', '_hrv'))

bio_cols = ['Average Pupil Dilation', 'Average Left Blink Rate', 'Average Right Blink Rate', 'RMSSD', 'SDNN']
available = [c for c in bio_cols if c in merged.columns]
bio_data = merged[available].dropna()

# --- FDR correction ---
n = len(available)
p_values = []
pairs = []
for i in range(n):
    for j in range(i+1, n):
        r, p = stats.spearmanr(bio_data[available[i]], bio_data[available[j]])
        p_values.append(p)
        pairs.append((available[i], available[j], r))

# Benjamini-Hochberg
m = len(p_values)
sorted_idx = np.argsort(p_values)
fdr_corrected = np.zeros(m)
for rank, idx in enumerate(sorted_idx):
    fdr_corrected[idx] = p_values[idx] * m / (rank + 1)
fdr_corrected = np.minimum.accumulate(fdr_corrected[np.argsort(sorted_idx)][::-1])[::-1]
fdr_corrected = np.clip(fdr_corrected, 0, 1)

print("=== Cross-Modal Correlations with FDR Correction ===\n")
print(f"{'Pair':<55} {'rho':>6} {'p_raw':>8} {'p_FDR':>8} {'sig':>4}")
for k, (v1, v2, r) in enumerate(pairs):
    sig = "*" if fdr_corrected[k] < 0.05 else "ns"
    print(f"{v1} vs {v2:<25} {r:>6.3f} {p_values[k]:>8.4f} {fdr_corrected[k]:>8.4f} {sig:>4}")

# --- regression: biometrics predicting score ---
score_col = 'Score'
if score_col not in merged.columns:
    score_col = 'Score_hrv' if 'Score_hrv' in merged.columns else None

if score_col:
    reg_data = merged[available + [score_col]].dropna()
    reg_data[score_col] = pd.to_numeric(reg_data[score_col], errors='coerce')
    reg_data = reg_data.dropna()
    
    X = reg_data[available].values
    y = reg_data[score_col].values
    
    # standardize
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    
    # fit regression
    model = LinearRegression()
    model.fit(X_scaled, y)
    y_pred = model.predict(X_scaled)
    
    # R-squared
    ss_res = np.sum((y - y_pred) ** 2)
    ss_tot = np.sum((y - np.mean(y)) ** 2)
    r_squared = 1 - ss_res / ss_tot if ss_tot > 0 else 0
    
    # coefficient significance
    n_obs = len(y)
    n_feat = X_scaled.shape[1]
    mse = ss_res / (n_obs - n_feat - 1)
    var_coef = mse * np.linalg.inv(X_scaled.T @ X_scaled).diagonal()
    se_coef = np.sqrt(np.abs(var_coef))
    t_stats = model.coef_ / se_coef
    p_coef = 2 * (1 - stats.t.cdf(np.abs(t_stats), df=n_obs - n_feat - 1))
    
    print(f"\n=== Regression: Biometrics -> Psychometric Score ===\n")
    print(f"R² = {r_squared:.4f}, n = {n_obs}")
    print(f"\n{'Feature':<35} {'Beta':>8} {'SE':>8} {'t':>8} {'p':>8}")
    for i, feat in enumerate(available):
        sig = "*" if p_coef[i] < 0.05 else ""
        print(f"{feat:<35} {model.coef_[i]:>8.4f} {se_coef[i]:>8.4f} {t_stats[i]:>8.3f} {p_coef[i]:>8.4f} {sig}")
    
    # actual vs predicted
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    axes[0].scatter(y, y_pred, alpha=0.5, s=20)
    axes[0].plot([y.min(), y.max()], [y.min(), y.max()], 'r--')
    axes[0].set_xlabel('Actual Score')
    axes[0].set_ylabel('Predicted Score')
    axes[0].set_title(f'Actual vs Predicted (R²={r_squared:.3f})')
    
    # coefficient bar chart
    coef_df = pd.DataFrame({'Feature': available, 'Beta': model.coef_, 'p': p_coef})
    coef_df = coef_df.sort_values('Beta')
    colors = ['tab:red' if p < 0.05 else 'tab:gray' for p in coef_df['p']]
    axes[1].barh(coef_df['Feature'], coef_df['Beta'], color=colors)
    axes[1].set_xlabel('Standardized Beta')
    axes[1].set_title('Regression Coefficients\n(red = p<.05)')
    axes[1].axvline(0, color='black', linewidth=0.5)
    
    plt.tight_layout()
    plt.show()
    plt.close()
else:
    print("\nNo Score column found for regression")
```
