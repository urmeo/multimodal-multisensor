```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA = '../data/individual/processed'

# load eye data
sed_data_01 = pd.read_csv(f'{DATA}/sed_fix_01.csv')
sed_data_02 = pd.read_csv(f'{DATA}/sed_fix_02.csv')
sed_data_03 = pd.read_csv(f'{DATA}/sed_fix_03.csv')

# session labels
sed_data_01['session'] = 'Session 01'
sed_data_02['session'] = 'Session 02'
sed_data_03['session'] = 'Session 03'

# combine datasets
all_sed_data = pd.concat([sed_data_01, sed_data_02, sed_data_03], ignore_index=True)

# plot fixation comparison
plt.figure(figsize=(14, 7))
sns.boxplot(data=all_sed_data, x='session', y='duration')
plt.title('Comparison of Fixation Durations Across Sessions')
plt.xlabel('Session')
plt.ylabel('Fixation Duration (s)')
plt.grid(True)
plt.tight_layout()
plt.show()
plt.close()

# descriptive statistics
fixation_duration_stats = all_sed_data[['session', 'duration']].groupby('session').describe()
fixation_duration_stats
```
