```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

OUTPUT = '../data/group_results'

# load pupil data
results_df = pd.read_csv(f'{OUTPUT}/Pupil_Dilation_STD.csv')

print(results_df)

# plot standard deviations
ax = results_df.set_index('Participant').plot(figsize=(15, 8), marker='o')

plt.title('Standard Deviation of Pupil Dilation for Each Participant Across Three Sessions')
plt.ylabel('Standard Deviation of Pupil Dilation')
plt.xlabel('Participant')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# reference lines
ax.axhline(y=0.1, color='blue', linestyle=':', label='Normal')
ax.axhline(y=0.2, color='purple', linestyle=':', label='Moderate Variation')
ax.axhline(y=0.3, color='green', linestyle=':', label='High Variation')

# merge legend handles
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles, labels, title='Session / Pupil Dilation Levels')
plt.show()
plt.close()
```
