```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

OUTPUT = '../data/group_results'

# load duration data
duration_df = pd.read_csv(f'{OUTPUT}/Psychometric_Test_Duration_STD.csv')

participants = duration_df['Participant'].tolist()

# build stats dict
stats = {}
for _, row in duration_df.iterrows():
    pid = row['Participant']
    session_data = pd.DataFrame({
        'Session': ['S01', 'S02', 'S03'],
        'STD of Duration (s)': [row['Session 01'], row['Session 02'], row['Session 03']]
    })
    stats[pid] = session_data

# plot STD duration
def plot_std_duration(stats, title):
    fig, ax = plt.subplots()
    colors = mpl.colormaps['tab10']
    
    for i, p_str in enumerate(stats):
        x = stats[p_str]['Session']
        y_std = stats[p_str]['STD of Duration (s)']
        ax.plot(x, y_std, marker='o', label=f'{p_str} - STD of Duration (s)', color=colors(i))
    
    plt.title(title)
    plt.ylabel('Time (seconds)')
    plt.xlabel('Session')
    plt.legend()
    plt.grid(True)
    plt.show()
    plt.close()

plot_std_duration(stats, 'STD of Duration by Session (10 Participants)')

for p_str in stats:
    print(f"Statistics for {p_str}:\n", stats[p_str])
```
